```python
import sqlite3
import pandas as pd

db_path = "../data/FPA_FOD_20170508.sqlite"   
conn = sqlite3.connect(db_path)

query = """
SELECT *
FROM Fires
WHERE FIRE_YEAR BETWEEN 2005 AND 2015
"""
df = pd.read_sql(query, conn)

conn.close()

output_path = "../data/wildfires_2005_2015.csv"
df.to_csv(output_path, index=False)

print("CSV exported successfully")
print(f"Rows exported: {len(df)}")
```

```python
df_wildfire = pd.read_csv("../data/wildfires_2005_2015.csv")

useful_cols = [
    "OBJECTID", 
    "FOD_ID",
    "FIRE_NAME",
    "FIRE_YEAR",
    "DISCOVERY_DATE",
    "DISCOVERY_DOY",
    "DISCOVERY_TIME",
    "STAT_CAUSE_CODE",
    "STAT_CAUSE_DESCR",
    "CONT_DATE",
    "CONT_DOY",
    "FIRE_SIZE",
    "FIRE_SIZE_CLASS",
    "LATITUDE",
    "LONGITUDE",
    "STATE",
    "COUNTY",
    "OWNER_DESCR"
]
df_wildfire_trimmed = df_wildfire[useful_cols]

print("Columns:", df_wildfire_trimmed.shape[1])

df_weather = pd.read_csv("../data/US_wildfire_weather_data.cvs")

df_merged = pd.merge(df_wildfire_trimmed, df_weather,on='OBJECTID',how='left')

df_merged.to_csv("../data/wildfire_and_weather_2005_2015.csv", index=False)

print("Merged CSV saved successfully")
print("Rows:", len(df_merged))
print("Columns:", df_merged.shape[1])
```


###### FOD_ID – Unique identifier for each fire incident.
###### FIRE_NAME – Assigned name of the wildfire.
###### FIRE_YEAR – Year when the fire was discovered.
###### DISCOVERY_DATE = Date on which the fire was discovered or confirmed to exist.
###### DISCOVERY_DOY – Day of the year the fire was discovered (1–365).
###### DISCOVERY_TIME – Time of fire discovery in HHMM format.
###### STAT_CAUSE_CODE = Code for the (statistical) cause of the fire.
###### STAT_CAUSE_DESCR = Description of the (statistical) cause of the fire.
###### CONT_DATE – Date when the fire was contained (Julian/serial format).
###### CONT_DOY – Day of the year the fire was contained.
###### FIRE_SIZE – Total acreage burned by the fire.
###### FIRE_SIZE_CLASS – Category of fire size based on acres burned.
###### LATITUDE – Latitude coordinate of the fire’s origin.
###### LONGITUDE – Longitude coordinate of the fire’s origin.
###### STATE – Two-letter U.S. state code where the fire occurred.
###### COUNTY – County identifier/FIPS county code.
###### OWNER_DESCR – Land ownership type where the fire originated.
